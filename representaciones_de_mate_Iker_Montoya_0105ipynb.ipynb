#### problema 1
Paso 1: Explicación paso a paso Problema 1: La suma de un número más 9 es igual a 21Definición de variable: Sea $x$ el número desconocido.Planteamiento: Traducimos el lenguaje común al algebraico:$$x + 9 = 21$$Despeje: Para aislar $x$, restamos 9 en ambos lados de la igualdad:$$x = 21 - 9$$$$x = 12$$Solución: El número es 12.

```python
# Definimos la ecuación: x + 9 = 21 -> x = 21 - 9
# resultado_total = 21
# numero_conocido = 9
resultado_total = int(input("Ingrese el resultado total: "))
numero_conocido = int(input("Ingrese el número conocido: "))

# despejamos: numero incognito = resultado final , numero conocido
resultado = resultado_total - numero_conocido



print(f"El número buscado es: {resultado}")
```

#### problema 2
Problema 2: Edades de Pedro y AnaDefinición de variables:Sea $a$ la edad de Ana.Como Pedro tiene el doble, su edad es $2a$.Planteamiento: La suma de sus edades es 36:$$a + 2a = 36$$Reducción y despeje:Sumamos términos semejantes: $3a = 36$Dividimos entre 3: $a = \frac{36}{3}$$a = 12$Cálculo final:Ana tiene $a = 12$ años.Pedro tiene $2a = 2 \times 12 = 24$ años.Solución: Ana tiene 12 años y Pedro tiene 24 años.

```python
# Definimos la suma: Ana(x) + Pedro(2x) = 36
#suma total = 36
int(input("Ingrese la suma total: "))

edad_ana = 36 / 3
edad_pedro = edad_ana * 2

print(f"Edad de Ana: {int(edad_ana)} años")
print(f"Edad de Pedro: {int(edad_pedro)} años")
```

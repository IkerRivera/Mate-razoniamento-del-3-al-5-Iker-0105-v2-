#### problema 3
Planteamiento del problema:"Un número": Como no sabemos cuál es, lo llamamos $x$."Que se le resta siete": Esto significa quitar 7 a nuestro número, es decir: $x - 7$."Da como resultado 18": Esto representa el signo de igualdad ($=$) y el valor final.La Ecuación:Juntando las partes anteriores, obtenemos la ecuación:$$x - 7 = 18$$Despeje de la variable:El objetivo es dejar a la $x$ sola de un lado del signo igual. Como el 7 está restando, aplicamos la operación inversa (la suma) en ambos lados de la igualdad para mantener el equilibrio:$$x = 18 + 7$$Resolución:Realizamos la operación:$$x = 25$$

```python
# Solicitamos al usuario que ingrese los valores conocidos
resultado_final = float(input("Ingresa el resultado de la resta : "))
numero_restado = float(input("Ingresa el número que se restó : "))

# Planteamos la ecuación: x - numero_restado = resultado_final
# Despejando: x = resultado_final + numero_restado
numero_incognito = resultado_final + numero_restado

# Mostramos el resultado
print(f"El número buscado es: {numero_incognito}")
```

#### problema 4
Explicación matemática: El triple de un número más 4 es igual a 31Para resolver este problema, debemos traducir el lenguaje natural a una estructura algebraica y luego despejar la incógnita paso a paso.Planteamiento del problema:"El triple de un número": Se representa como $3x$."Más 4": Se representa como $+ 4$."Es igual a 31": Se representa como $= 31$.La Ecuación:Combinamos los elementos:$$3x + 4 = 31$$Despeje de la variable:Debemos aislar $x$ siguiendo el orden jerárquico inverso de las operaciones:Paso A (Eliminar la suma): Restamos 4 en ambos lados para mover el término independiente.$$3x = 31 - 4$$$$3x = 27$$Paso B (Eliminar la multiplicación): Como el 3 está multiplicando a la $x$, dividimos ambos lados entre 3.$$x = \frac{27}{3}$$$$x = 9$$

```python
# Solicitamos los datos al usuario
resultado_igualdad = float(input("¿A qué es igual la ecuación? : "))
valor_sumado = float(input("¿Qué valor se suma? : "))
multiplicador = float(input("¿Por cuánto se multiplica el número? : "))

# Planteamiento: (multiplicador * x) + valor_sumado = resultado_igualdad
# Despeje paso a paso:
# 1. Restamos el valor sumado: resultado_igualdad - valor_sumado
# 2. Dividimos por el multiplicador: (resultado_igualdad - valor_sumado) / multiplicador

numero_incognito = (resultado_igualdad - valor_sumado) / multiplicador

# Mostramos el resultado
print(f"El número buscado es: {numero_incognito}")
```

#### problema 5
Explicación matemática: Dimensiones de un terreno rectangularPara resolver este problema, debemos recordar que el perímetro de un rectángulo es la suma de todos sus lados, o bien, la fórmula:$$P = 2 \times (\text{largo} + \text{ancho})$$Definición de variables:Sea $x$ el ancho del terreno (en metros).El largo mide 5 metros más que el ancho, por lo tanto, el largo es $(x + 5)$.Planteamiento de la ecuación:Sustituimos nuestros valores en la fórmula del perímetro:$$50 = 2 \times ((x + 5) + x)$$Despeje y resolución:Primero, simplificamos lo que está dentro del paréntesis:$$50 = 2 \times (2x + 5)$$Ahora, podemos dividir ambos lados de la ecuación entre 2 para simplificar:$$25 = 2x + 5$$Restamos 5 en ambos lados:$$20 = 2x$$Finalmente, dividimos entre 2:$$x = 10$$Cálculo de los lados:Ancho: $x = 10$ metros.Largo: $x + 5 = 10 + 5 = 15$ metros.

```python
# Solicitamos los datos al usuario
perimetro = float(input("Ingresa el perímetro total : "))
diferencia = float(input("¿Cuánto más mide el largo que el ancho? : "))

# Ecuación deducida: 2 * (x + (x + diferencia)) = perimetro
# Simplificada: 2 * (2x + diferencia) = perimetro
# Entonces: x = ( (perimetro / 2) - diferencia ) / 2

ancho = ((perimetro / 2) - diferencia) / 2
largo = ancho + diferencia

print(f"El ancho del terreno es: {ancho} metros")
print(f"El largo del terreno es: {largo} metros")
```
